```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/iris.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df["target"].value_counts()
```

```python
df["target"].value_counts().plot(kind="bar")
plt.xlabel("Target Class")
plt.ylabel("Number of Samples")
plt.title("Target Class Distribution")
plt.show()
```

```python
df.boxplot(column="petal length (cm)", by="target")
plt.title("Petal Length by Target Class")
plt.suptitle("")
plt.xlabel("Target Class")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
df.boxplot(column="petal width (cm)", by="target")
plt.title("Petal Width by Target Class")
plt.suptitle("")
plt.xlabel("Target Class")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python
### Observations

"""- The target classes are balanced, with 50 samples in each class.
- Petal length shows noticeable differences between the target classes.
- Petal width also shows noticeable differences between the target classes.
- These features may be useful for distinguishing the three classes."""
```

```python
## Conclusion

"""The Iris dataset contains 150 samples and 4 numerical features.
There are 3 target classes with 50 samples each, so the target distribution is balanced.

The exploratory analysis shows that petal length and petal width differ across the target classes.
These features appear useful for distinguishing between the classes and can be investigated further during model development."""
```
